```python

import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")
df =titanic.loc[:,["age","class","sex","fare","survived"]]
df.head()
```

```python
pd.pivot_table(
    df,
    index = "class",
    columns= "sex",
    values= "age",
    aggfunc= "mean"
)
```

```python
pd.pivot_table(
    df,
    index = "class",
    columns="sex",
    values="survived",
    aggfunc = ["mean","sum"]
)
```
