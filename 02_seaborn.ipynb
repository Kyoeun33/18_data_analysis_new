```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")

titanic.head()
```

```python
#회귀선이 있는 산점도

import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1,2,figsize = (15,5))

sns.regplot(
    x = "age",
    y="fare", 
    data = titanic,
    ax=axes[0]
)

sns.regplot(
    x = "age",
    y="fare", 
    data = titanic,
    ax=axes[1],
    fit_reg=False

)

plt.show()
```

```python

# 히스토드램 / 커널 밀도 함수



fig,axes = plt.subplot_mosaic([['top_left','top_right'],
                    ["middle_left","middle_right"],
                    ['bottom','bottom']],
                    figsize = (15,6),
                    constrained_layout = True)

sns.histplot(
    x="age",
    data = titanic,
    ax = axes['top_left']
)

sns.histplot(
    x="age",
    data = titanic,
    multiple='dodge',
    hue='survived',
    ax = axes['top_right']
)

sns.histplot(
    x="age",
    data = titanic,
    multiple='stack',
    hue='survived',
    ax = axes['middle_left']
)

sns.histplot(
    x="age",
    data = titanic,
    multiple='fill',
    hue='survived',
    ax = axes['middle_right']
)

sns.histplot(
    x="age",
    data = titanic,
    multiple='fill',
    bins=10,
    hue='survived',
    ax = axes['bottom']
)

fig.suptitle('titanic - Age Distribution')

axes["top_left"].set_title('Histogram')
axes["top_right"].set_title('Histogram(hue)')
axes["middle_left"].set_title('Histogram(dodge)')
axes["middle_left"].set_title('Histogram(stack)')
axes["bottom"].set_title('Histogram(fill)')


plt.show()
```

```python
fig,axes =plt.subplot_mosaic(
    [
        ["top_left", "top_center", "right"],
        ["bottom_left", "bottom_center", "right"],
    ],
    figsize=(15, 6),
    constrained_layout=True
)

sns.kdeplot(
    x="age",
    data=titanic,
    ax=axes["top_left"]
)

sns.kdeplot(
    x="age",
    data=titanic,
    hue="survived",
    ax=axes["bottom_left"]
)

sns.kdeplot(
    x="age",
    data=titanic,
    hue="survived",
    fill=True,
    ax=axes["top_center"]
)

sns.kdeplot(
    x="age",
    data=titanic,
    multiple="stack",
    hue="survived",
    ax=axes["bottom_center"]
)

sns.kdeplot(
    x="age",
    data=titanic,
    multiple="fill",
    bw_adjust=2.0,
    hue="survived",
    ax=axes["right"]
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset("titanic")
sns.set_style("darkgrid")

table = titanic.pivot_table(
    index=["sex"],
    columns=["class"],
    aggfunc = "size",
    observed = False
)

print(table)
type(table)
```

```python
sns.heatmap(
    table,
    annot = True,
    fmt = "d",
    cmap = "YlGnBu",
    linewidth = 5,
    cbar = True
)

plt.show()
```

```python
# 범주형 데이터의 산점도
sns.set_style("whitegrid")

fig, axes = plt.subplots(
    1, 2, figsize=(15, 5)
)

sns.stripplot(
    x="class", 
    y="age", 
    data=titanic, 
    ax=axes[0]
)

sns.stripplot(
    x="class", 
    y="age", 
    data=titanic, 
    ax=axes[1], 
    hue="class",
    size=3

)

axes[0].set_title("Strip plot")

plt.show()
```
