```python
dict_data ={
    "c0":[1, 2, 3], 
    "c1":[4, 5, 6],
    "c2":[7, 8, 9],
    "c3":[10, 11, 12],
    "c4":[13, 14, 15],

}

df = pd.DataFrame(dict_data,index = ["r0","r2","r1"])
df
```

```python
ndf = df.sort_index(ascending=False) # ascemding = 오름차순
ndf
```
