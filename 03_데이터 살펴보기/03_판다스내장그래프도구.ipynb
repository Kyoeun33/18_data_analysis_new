```python
import pandas as pd
df = pd.read_excel("./data/남북한발전전력량.xlsx")

df_ns =df.iloc[[0,5],3:]

df_ns.index = ["south","North"]

df_ns = df_ns.astype(int)

tdf_ns = df_ns.T
tdf_ns.plot(kind='hist')
```

```python
tdf_ns.plot(kind = "hist",
            alpha = 0.5, # 투명도
            figsize=(6,3), # 표 비율
            grid= True, # 눈금
            color = 'blue', 
            edgecolor = 'red'
            
            )
```
