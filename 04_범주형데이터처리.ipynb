```python
import pandas as pd
import numpy as np


df = pd.read_csv("./data/auto-mpg.csv",header=None)
df.columns=[
    "mpg","cyliders","displacement","horsepower","weight","acceleration","model year","origin","name"
]

df["horsepower"] = df["horsepower"].replace("?", np.nan)

df = df.dropna(subset=["horsepower"])
df["horsepower"] = df["horsepower"].astype("float")

df.head()
```

```python
df["horsepower"].describe()
```

```python
count, bin_dividers = np.histogram(df["horsepower"],bins=3)
print(count)
print(bin_dividers)
```

```python
bin_names = ["저출력", "보통출력", "고출력"]

df["hp_bins"]= pd.cut(
    x = df [ "horsepower"],
    bins = bin_dividers,
    labels=bin_names,
    include_lowest=True

)

df[["horsepower","hp_bins"]].head(10)
```
