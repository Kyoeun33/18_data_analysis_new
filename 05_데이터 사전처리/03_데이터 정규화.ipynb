```python
import pandas as pd

df = pd.read_csv("./data/auto-mpg.csv",header=None)
df.columns=[
    "mpg","cyliders","displacement","horsepower","weight","acceleration","model year","origin","name"
]

df
```

```python
mpg_to_kpl = 1.60934 / 3.78541

df["kpl"] = df["mpg"] * mpg_to_kpl

df
```

```python
df["kpl"] = df["kpl"].round(2)

df
```
