```python
import pandas as pd
import numpy as np


df = pd.read_csv("./data/auto-mpg.csv",header=None)
df.columns=[
    "mpg","cyliders","displacement","horsepower","weight","acceleration","model year","origin","name"
]

df["horsepower"] = df["horsepower"].replace("?", np.nan)

df = df.dropna(subset=["horsepower"])
df["horsepower"] = df["horsepower"].astype("float")
```

```python
df.head()
```

```python
df["horsepower"].describe()
```

```python
# min - max - scaling
##두가지 방법
#  min = 46.0 max =230.0

df["horsepower_minmax"] = (df["horsepower"] - df["horsepower"].min()) / (df["horsepower"].max()-df["horsepower"].min())

print(df["horsepower_minmax"])
print ((df["horsepower"]-46)/(230 - 46))
```

```python
df["horsepower_minmax"].describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df [ "horsepower_minmax"] = scaler.fit_transform(df[['horsepower']]) ##[[]] <== 데이터 프레임 형태

df["horsepower_minmax"]
```

```python
df["horsepower_minmax"].describe()
```

```python
# standard Scaling

df["horsepoer_standard"] = (df["horsepower"] - df['horsepower'].mean())/df["horsepower"].std()
```
