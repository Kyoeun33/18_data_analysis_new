```python
import pandas as pd
```

```python
dict_data = {
    "c1" : ["a","a","b",'a',"b"],
    "c2" : [1,1,1,2,3],
    "c3" : [1,1,2,2,2],
}

df = pd.DataFrame(dict_data)

df
```

```python
df.duplicated()
```

```python
df.duplicated(keep="last")
```

```python
df.duplicated(keep=False)
```

```python
df["c2"].duplicated()
```

```python
df.duplicated(subset=["c2"])
```
