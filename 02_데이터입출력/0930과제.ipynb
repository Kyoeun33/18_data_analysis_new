```python
# 0930퀴즈
# 데이터 프레임
# {
#     '이름': ['김철수', '이영희', '박민수', '최지우'],
#     '나이': [20, 21, 22, 20],
#     '성별': ['남', '여', '남', '여'],
#     '수학': [90, 85, 70, 100],
#     '영어': [85, 95, 75, 90]
# }

# 1. df 라는 이름의 데이터프레임을 만들어 주세요

import pandas as pd

df = {
    '이름': ['김철수', '이영희', '박민수', '최지우'],
    '나이': [20, 21, 22, 20],
    '성별': ['남', '여', '남', '여'],
    '수학': [90, 85, 70, 100],
    '영어': [85, 95, 75, 90]
}  

df = pd.DataFrame(df)
print(df)
# 2. 생성된 데이터프레임의 상위 2개 행만 출력하세요.
print(df.loc[0:1])

# 3. 데이터프레임의 요약 정보(컬럼명, 데이터 타입 )을 확인하세요
print(type(df))
# 4. '이름' 컬럼을 인덱스로 설정하세요. (기존 df를 수정)
name={'이름':['A',"B","c"]}
df = df.set_index(["이름"])
print(df)
# 5. loc을 사용하여 '이영희'의 '수학' 점수와 '영어' 점수만 추출하세요.
print()
print(df.loc["이영희",['수학','영어']])
print(df.loc["이영희",'수학':'영어'])
# 6. iloc을 사용하여 첫 번째 행부터 두 번째 행까지, 그리고 2번째 열(수학)부터 마지막 열까지 슬라이싱하세요.
print()
print(df.iloc[0:1,1:3])
print()

# 시리즈
sr = {
    '이름': ['김철수', '이영희', '박민수', '최지우'],
    '나이': [20, 21, 22, 20],
    '성별': ['남', '여', '남', '여'],
    '수학': [90, 85, 70, 100],
    '영어': [85, 95, 75, 90]
}

# 1. 위 데이터로 시리즈 객체 sr을 만들어주세요
sr = pd.Series(sr)

# 2. 시리즈의 인덱스 배열과 데이터값 배열을 각각 추출하여 화면에 출력하세요.
print(sr.index)
print(sr.values)


list_data = ["2010-01-02", 3.14, "ABC", 100, True]
# 3. 위 데이터를 시리즈 객체로 만들어 주세요
list_data = pd.Series(list_data)

# 4. 데이터 타입을 출력해보세요
print(type(list_data))

tup_data = ("영인", "2010-05-01", "여", True)
index_labels = ["이름", "생년월일", "성별", "학생여부"]
# 5. 위 데이터를 시리즈객체로 생성해주세요. 
tup_data = pd.Series(tup_data,index= index_labels)
index_labels = ["이름", "생년월일", "성별", "학생여부"]

index_labels = pd.Series(index_labels, index=index_labels)


# 6. 해당 시리즈의 원소 개수, 형상(shape), 차원(ndim) 을 출력해주세요 
print()
print(len(tup_data),tup_data.shape,tup_data.ndim)
print(len(index_labels),index_labels.shape,index_labels.ndim)


# 7. 위치 기반(정수)으로 첫 번째 원소를 안전하게 조회하는 코드를 작성하세요. (iloc 활용)

print()
print(tup_data.iloc[0])

# 8. 라벨 '이름'으로 원소를 조회하는 코드를 작성하세요.

print()
print(tup_data['이름'])

# 9. 위치 인덱스 번호 리스트를 활용하여 '생년월일'과 '성별'을 가져오세요. (iloc 권장)

print()
print(index_labels.iloc[1:3])

# 10. 라벨 이름 리스트를 활용하여 '생년월일'과 '성별'을 가져오세요.

print(tup_data.loc["생년월일": "성별"])

# ---
# 11. 아무 데이터도 전달하지 않은 기본형 빈 시리즈를 생성하세요.
print()
a=pd.Series()
print(a)

# 12. 스칼라 값 5와 인덱스 ["a", "b", "c"]를 사용하여 시리즈를 생성하고 출력하세요.

print()
print(pd.Series(5,index =["a", "b", "c"]))
```
