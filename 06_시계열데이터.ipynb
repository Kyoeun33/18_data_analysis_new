```python
# timestamp를 이용해서 period로 변환

dates = ["2019-01-01","2020-03-01","2021-06-01"]

ts_dates = pd.to_datetime(dates)
ts_dates
```

```python
pr_day = ts_dates.to_period(freq="D")
pr_day
```

```python
pr_month=ts_dates.to_period(freq = "M")
pr_month
```

```python
pr_year=ts_dates.to_period(freq = "Y")
pr_year
```

```python
#시계열 데이터 만들기
#월 시작일 기준
ts_ms = pd.date_range(
    start = "2024-01-01",
    end=None,
    periods=6,
    freq="MS",
    tz='Asia/Seoul'
)

ts_ms
```

```python
# 월 막말 기준
ts_me = pd.date_range(
    start = "2024-01-01",
    end=None,
    periods=6,
    freq="ME",
    tz='Asia/Seoul'
)

ts_me
```

```python
# 분기(3개월)간격
ts_3m = pd.date_range(
    start = "2024-01-01",
    periods=6,
    freq="3MS",
    tz='Asia/Seoul'
)

ts_3m
```

```python
# 시계열 데이터 활용

df = pd.read_csv("./data/stock-data.csv")
df.head()
```

```python
df["new_Date"]= pd.to_datetime(df["Date"])
df.head()
```

```python
df["new_Date"]
```

```python
df["year"] = df["new_Date"].dt.year
```

```python
df["month"] = df["new_Date"].dt.month
```

```python
df["day"] = df["new_Date"].dt.day
```

```python
df.head()
```

```python
df["Date_yr"] = df["new_Date"].dt.to_period(freq="Y")
```

```python
df["Date_mo"] = df["new_Date"].dt.to_period(freq="M")
df.info()
```
